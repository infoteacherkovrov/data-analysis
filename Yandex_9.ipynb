```python
#Соловьев
def main():
    num1 = 8465
    num2 = 9799
    product = num1 * num2  # 8465 * 9799
    max_value = 10_101_010_000  # до 10.1 млрд
    found_numbers = []

    # Маска: 10*2426?
    prefix = "10"
    mid = "2426"

    for x_len in range(0, 6):  # до 5 цифр между,больше нету смысла
        if x_len == 0:
            x_values = [""]
        else:
            x_values = [str(i).zfill(x_len) for i in range(10**x_len)]
        for x in x_values:
            for d in range(10):
                num_str = prefix + x + mid + str(d)
                if len(num_str) > 10:
                    continue
                num = int(num_str)
                if num > max_value:
                    continue
                found_numbers.append(num)

    found_numbers = sorted(set(found_numbers))
    answer = []

    for num in found_numbers:
        if num % product == 0:
            answer.append((num, num // product))
        elif num % num1 == 0:
            answer.append((num, num // num1))
        elif num % num2 == 0:
            answer.append((num, num // num2))

    for num, res in answer:
        print(num, res)

if __name__ == "__main__": #условное выполнение функции main, когда скрипт запускаеся напрямую
    main()
#было сложно
```

```python
from fnmatch import * # Баранов
def f(n):
    if n % (8465*9799) == 0:
        return n // (8465*9799)
    elif n % 8465 == 0:
        return n // 8465
    else:
        return n // 9799

res = []

for n in range(8465, 10**10+1, 8465):
    if fnmatch(str(n), '10*2426?'):
        res.append([n, f(n)])

for n1 in range(9799, 10**10+1, 9799):
    if fnmatch(str(n1), '10*2426?'):
        res.append([n1, f(n1)])

res.sort()

for i in range(len(res)):
    print(res[i][0], res[i][1])
```

```python
# Неэффективно, но хоть так
with open('26 (7).txt') as f:
    n,k=map(int,f.readline().split())
    chair=list((i+1,int(f.readline())) for i in range(n))
    player=list((i+1,int(f.readline())) for i in range(k))
ans=[]
for i in player:
    r=10**10
    for j in chair:
        if abs(i[1]-j[1])<r:
            r=abs(i[1]-j[1])
            p=(j,r,i)
    ans.append(p)
ans.sort()
ans2=[ans[0]]
for i in ans:
    if i[0]!=ans2[-1][0]:
        ans2.append(i)
last=0
lastn=0
for i in ans2:
    if i[1]>last:
        last=i[1]
        lastn=i[2][0]
    elif i[1]==last and i[2][0]>lastn:
        lastn=i[2][0]
print(k-len(ans2),lastn)
```
