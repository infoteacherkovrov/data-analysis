```python
#соловьев
def has_digit_2_or_3(n):
    return '2' in str(n) or '3' in str(n)
def is_prime(n):
    if n < 2:return False
    if n == 2:return True
    if n % 2 == 0:return False
    d = 3
    while d * d <= n:
        if n % d == 0:return False
        d += 2
    return True
def get_unique_prime_factors(n):
    factors = set()
    d = 2
    while d * d <= n:
        if n % d == 0:
            if not is_prime(d):return None
            factors.add(d)
            while n % d == 0:
                n //= d
        d += 1
    if n > 1:
        if is_prime(n):
            factors.add(n)
        else:return None
    return factors
# Поиск
count = 0
n = 5000001
results = []

while count < 5:
    primes = get_unique_prime_factors(n)
    if primes and len(primes) == 3:
        if all(has_digit_2_or_3(p) for p in primes):
            results.append(n)
            count += 1
    n += 1
for num in results:
    print(num)   #то что выводится ближе к ответу, но всё-равно не то
```
