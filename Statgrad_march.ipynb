```python
#соловьев
def has_digit_2_or_3(n):
    """Проверяет, содержит ли число хотя бы одну цифру 2 или 3"""
    s = str(n)
    return '2' in s or '3' in s
def is_prime(n):
    """Проверяет, является ли число простым"""
    if n < 2:return False
    if n == 2:return True
    if n % 2 == 0:return False
    d = 3
    while d * d <= n:
        if n % d == 0:return False
        d += 2
    return True
def get_prime_factors(n):
    """Возвращает список простых множителей (с повторениями)"""
    factors = []
    d = 2
    while d * d <= n:
        while n % d == 0:
            if is_prime(d):
                factors.append(d)
            n //= d
        d += 1
    if n > 1:
        factors.append(n)
    return factors
def satisfies_condition(n):
    """Проверяет, что n — произведение ровно трёх простых множителей, каждый содержит 2 или 3"""
    factors = get_prime_factors(n)
    if len(factors) != 3:
        return False
    for p in factors:
        if not is_prime(p) or not has_digit_2_or_3(p):
            return False
    return True
# Перебор чисел, начиная с 5_000_001
count = 0
n = 5000001
results = []
while count < 5:
    if satisfies_condition(n):
        results.append(n)
        count += 1
    n += 1
for num in results:
    print(num)
```

# Задание 24

```python
import string #Баранов
s = open('24.txt').readline()
for elem in '0123456789':
    s = s.replace(elem, '*')
while '**' in s:
    s = s.replace('**', '*')
maxa = 0
def alld(s):
    alf = string.ascii_uppercase
    for elem in alf:
        if elem not in s:
            return False
    return True
for l in range(len(s)):
    for r in range(l+maxa, len(s)):
        t = s[l:r+1]
        if '*' in t:
            break
        if alld(t):
            maxa = max(maxa, len(t))
print(maxa)
```
